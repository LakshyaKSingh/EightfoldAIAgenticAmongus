```python
import numpy as np

risk_features = [
    "institution_risk_score",
    "gpa_anomaly_score",
    "company_risk_score",
    "tenure_gap_months",
    "applications_7d",
    "applications_30d",
    "app_to_avg_ratio",
    "time_since_last_app_hrs",
    "is_new_device",
    "failed_logins_24h",
    "login_velocity_24h",
    "email_risk_score",
    "ip_risk_score",
    "profile_age_days",
    "copy_paste_ratio",
    "skills_to_exp_ratio"
]

z_scores = {}

for col in risk_features:
    mean = df[col].mean()
    std = df[col].std()
    z_scores[col] = (df[col] - mean) / std

z_df = pd.DataFrame(z_scores)
```

```python
z_df
```

```python
df["credential_score"] = (
    z_df["institution_risk_score"] +
    z_df["gpa_anomaly_score"] +
    z_df["company_risk_score"] +
    z_df["tenure_gap_months"]
)
```

```python
df["bombing_score"] = (
    z_df["applications_7d"] +
    z_df["applications_30d"] +
    z_df["app_to_avg_ratio"] -
    z_df["time_since_last_app_hrs"]
)
```

```python
df["ato_score"] = (
    z_df["is_new_device"] +
    z_df["failed_logins_24h"] +
    z_df["login_velocity_24h"] +
    z_df["email_risk_score"] +
    z_df["ip_risk_score"]
)
```

```python
df["ghost_score"] = (
    -z_df["profile_age_days"] +
    z_df["copy_paste_ratio"] +
    z_df["skills_to_exp_ratio"]
)
```

```python
score_cols = ["credential_score", "bombing_score", "ato_score", "ghost_score"]

df["predicted_type"] = df[score_cols].idxmax(axis=1)
```

```python
df.loc[df["labels"] == 0, "predicted_type"] = "genuine"
```

```python
df["applications_30d"] = np.log1p(df["applications_30d"])
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.metrics import roc_auc_score
from sklearn.ensemble import RandomForestClassifier

score_cols = [
    "credential_score",
    "bombing_score",
    "ato_score",
    "ghost_score"
]

# =========================
# 1️⃣ Grouped Statistics
# =========================
print("\n===== GROUPED DESCRIPTIVE STATS =====\n")
for col in score_cols:
    print(f"\n--- {col} ---")
    print(df.groupby("labels")[col].describe())


# =========================
# 2️⃣ KDE Plots
# =========================
plt.figure(figsize=(12, 8))
for i, col in enumerate(score_cols, 1):
    plt.subplot(2, 2, i)
    sns.kdeplot(data=df, x=col, hue="labels", common_norm=False)
    plt.title(f"KDE: {col}")
plt.tight_layout()
plt.show()


# =========================
# 3️⃣ ROC-AUC Per Score
# =========================
print("\n===== ROC-AUC SCORES =====\n")
for col in score_cols:
    auc = roc_auc_score(df["labels"], df[col])
    print(f"{col}: {auc:.4f}")


# =========================
# 4️⃣ Normalize Scores
# =========================
scaler = MinMaxScaler()
df[score_cols] = scaler.fit_transform(df[score_cols])

print("\nScores normalized using MinMaxScaler.")


# =========================
# 5️⃣ Fraud-Type Purity
# =========================
print("\n===== FRAUD TYPE PURITY =====\n")
purity = df.groupby("predicted_type")["labels"].mean()
print(purity)


# =========================
# 6️⃣ Feature Importance Check
# =========================
print("\n===== RANDOM FOREST FEATURE IMPORTANCE =====\n")

risk_features = [
    "institution_risk_score",
    "gpa_anomaly_score",
    "company_risk_score",
    "tenure_gap_months",
    "applications_7d",
    "applications_30d",
    "app_to_avg_ratio",
    "time_since_last_app_hrs",
    "is_new_device",
    "failed_logins_24h",
    "login_velocity_24h",
    "email_risk_score",
    "ip_risk_score",
    "profile_age_days",
    "copy_paste_ratio",
    "skills_to_exp_ratio"
]

rf = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42
)

rf.fit(df[risk_features], df["labels"])

importances = pd.Series(
    rf.feature_importances_,
    index=risk_features
).sort_values(ascending=False)

print(importances)

plt.figure(figsize=(8,6))
importances.plot(kind="barh")
plt.title("Feature Importance")
plt.gca().invert_yaxis()
plt.show()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import roc_auc_score

# -----------------------------
# 1️⃣ Define Risk Features
# -----------------------------
risk_features = [
    "institution_risk_score",
    "gpa_anomaly_score",
    "company_risk_score",
    "tenure_gap_months",
    "applications_7d",
    "applications_30d",
    "app_to_avg_ratio",
    "time_since_last_app_hrs",
    "is_new_device",
    "failed_logins_24h",
    "login_velocity_24h",
    "email_risk_score",
    "ip_risk_score",
    "profile_age_days",
    "copy_paste_ratio",
    "skills_to_exp_ratio"
]

X = df[risk_features].copy()
y = df["labels"]

# -----------------------------
# 2️⃣ Standardize Features
# -----------------------------
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# -----------------------------
# 3️⃣ KMeans Clustering
# -----------------------------
kmeans = KMeans(n_clusters=8, random_state=42)
clusters = kmeans.fit_predict(X_scaled)

df["cluster"] = clusters

# Fraud density per cluster
cluster_stats = df.groupby("cluster")["labels"].agg(["mean", "count"])\
                  .sort_values("mean", ascending=False)

print("\n===== Cluster Fraud Density =====\n")
print(cluster_stats)

# -----------------------------
# 4️⃣ Identify High-Fraud Clusters
# -----------------------------
fraud_clusters = cluster_stats[cluster_stats["mean"] > 0.8].index.tolist()

df["cluster_flag"] = df["cluster"].apply(
    lambda x: 1 if x in fraud_clusters else 0
)

print("\nHigh Fraud Clusters:", fraud_clusters)

# -----------------------------
# 5️⃣ Train Nonlinear Classifier
#    (Include cluster_flag as feature)
# -----------------------------
model_features = risk_features + ["cluster_flag"]

X_model = df[model_features]

model = GradientBoostingClassifier(random_state=42)
model.fit(X_model, y)

probs = model.predict_proba(X_model)[:, 1]
df["fraud_prob"] = probs

auc = roc_auc_score(y, probs)
print(f"\nModel ROC-AUC: {auc:.4f}")

# -----------------------------
# 6️⃣ PCA for Visualization
# -----------------------------
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

df["pca1"] = X_pca[:, 0]
df["pca2"] = X_pca[:, 1]

# -----------------------------
# 7️⃣ Visualize Clusters
# -----------------------------
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x="pca1",
    y="pca2",
    hue="cluster",
    palette="tab10",
    data=df,
    alpha=0.6
)
plt.title("KMeans Clusters (PCA Projection)")
plt.show()

# -----------------------------
# 8️⃣ Visualize True Labels
# -----------------------------
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x="pca1",
    y="pca2",
    hue="labels",
    palette=["blue", "red"],
    data=df,
    alpha=0.6
)
plt.title("True Fraud Labels (PCA Projection)")
plt.show()

# -----------------------------
# 9️⃣ Visualize Fraud Probability Heatmap
# -----------------------------
plt.figure(figsize=(10, 6))
plt.scatter(
    df["pca1"],
    df["pca2"],
    c=df["fraud_prob"],
    cmap="coolwarm",
    alpha=0.6
)
plt.colorbar(label="Fraud Probability")
plt.title("Fraud Probability Heatmap (Model Output)")
plt.show()
```
